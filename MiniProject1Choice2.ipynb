## The task
Write a python script that 

1. Calculates the first 200 terms of the Fibonacci sequence 
$$
f_n = f_{n-1} + f_{n-2},
$$
with
$f_0 =$ 0 and $f_1 = $ 1.
The calculation is carried out using either 32-bits integers, integers and double-precision floats.
3. Stores the result of the three sequences, together with the value of $n$, in an array for $n = 20, 40,\dots, 200$. 
4. Saves the array to a file.

## Description of the python code

The code is similar to that of Guided Example 1: you should inspect that before continuing.

* To calculate the terms of the sequence use a `for` loop. The number of terms of the sequence is given by the variable `N`. 
* The sequence is calculated using either 32-bits integers, integers and double-precision floats. Arrays of dimension N and of data-type 32-bits integers, 64-bits integers and 64-bits double-precision floats are assigned to variables `f`, `g` and `h`. 
* The array that stores the results is assigned to the variable `results`. 
* When the number of terms `n` in the sum is a multiple of 20 (the variable `M`), `n`and the nth term of `f`, `g`, and `h` is stored in the first, second, third and fourth column of `results`.
* The array `results` is written onto the file `table.csv`. The delimiter (character that separates the columns) is set to `,` so that the resulting file is in the comma separated value format. The `fmt` (the format of the columns) is set to `%d. %d,%d,%.16e`. 



## Running the code
Hitting run should create the `table.csv` file that should appear in the files list of this repl. To see if the results are what you expect click on the file. If the file content is what you expect, then follow the instructions on Canvas for the report for this assignment.

# Python code


```python
import numpy as np

#Setting variables
N       = 201
M       = 20
found32 = False
found64 = False
results = np.zeros((int(N/M)+1,4))
f       = np.zeros(N,np.int32)
g       = np.zeros(N,np.int64)
h       = np.zeros(N,np.float64)

#Set initial values
f[0], f[1] = 0, 1
g[0], g[1] = 0, 1
h[0], h[1] = 0.0, 1.0

#Get first 200 values in sequence
for n in range(2, N):
    f[n] = f[n-1] + f[n-2]
    g[n] = g[n-1] + g[n-2] 
    h[n] = h[n-1] + h[n-2] 

    #Get every 20th value in sequence
    if n%M == 0:
        row = n // M
        
        results[row, 0] = n
        results[row, 1] = f[n]
        results[row, 2] = g[n]
        results[row, 3] = h[n]

    #Finding term where data type becomes unreliable
    if f[n] != g[n] and found32 == False:
        print(n)
        found32 = True

    if g[n] != h[n] and found64 == False:
        print(n)
        found64 = True

with open('table.csv', 'w') as f:
    
    #Adding column headers
    f.write('Term Number, 32-bit Integers, 64-bit Integers, Double-Precision Floats\n')
    
    #Save data
    np.savetxt(f, results, delimiter=',', fmt='%d, %d, %d, %.16e')
```
